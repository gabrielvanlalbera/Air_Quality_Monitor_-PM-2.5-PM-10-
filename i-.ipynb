```python
import pandas as pd
import os

# Load PM dataset
df = pd.read_csv("pm1.csv")

image_folder = "images/"

data = []

for i in range(48):  # 0 to 47
    
    # Image time in seconds
    image_time = (i + 1) * 60
    
    # Get PM values within that 60 sec window
    pm_window = df[(df["Time_s"] > image_time - 60) &
                   (df["Time_s"] <= image_time)]
    
    if not pm_window.empty:
        data.append({
            "image_path": os.path.join(image_folder, f"{i}.jpg"),
            "pm2_5": pm_window["PM2.5"].mean(),
            "pm10": pm_window["PM10"].mean()
        })

final_df = pd.DataFrame(data)

print(final_df.head())
```
